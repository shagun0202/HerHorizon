```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("PCOS_data_without_infertility.xlsx", sheet_name="Full_new")
```

```python
sns.heatmap(df.isnull(),yticklabels=False,cbar=False,cmap='viridis')
```

```python
print(df.columns.tolist())
```

```python
df.columns = df.columns.str.strip()
```

```python
df = df.drop(columns=['Unnamed: 44'])
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(r'[^\w\s]', '', regex=True)   # remove . ( ) / : etc.
    .str.replace(r'\s+', '_', regex=True)      # spaces -> underscore
)
```

```python
df['marraige_status_yrs'] = df['marraige_status_yrs'].fillna(df['marraige_status_yrs'].median())
```

```python
fast_food_mode = df['fast_food_yn'].mode()[0]
df['fast_food_yn'] = df['fast_food_yn'].fillna(fast_food_mode)
```

```python
df.isnull().sum()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(15,12))

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

# Feature Engineering

```python
df = df.drop(columns=["sl_no", "patient_file_no"])
```

```python
df[["weight_kg", "heightcm", "bmi"]].head()
```

```python
df["Age_Group"] = pd.cut(
    df["age_yrs"],
    bins=[0, 18, 25, 35, 100],
    labels=["<18", "18-25", "26-35", "35+"]
)
```

```python
df["Age_Group"].value_counts()
```

```python
df["Long_Cycle"] = (
    df["cycle_lengthdays"] > 35
).astype(int)
```

```python
symptom_cols = [
    "hair_growthyn",
    "pimplesyn",
    "hair_lossyn",
    "weight_gainyn",
    "skin_darkening_yn"
]

df["Symptom_Score"] = df[symptom_cols].sum(axis=1)
```

```python
lifestyle_cols = [
    "fast_food_yn",
    "regexerciseyn"
]
```

```python
for col in lifestyle_cols:
    print(col)
    print(df[col].value_counts())
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x="Age_Group"
)

plt.title("Number of Patients by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")
for container in ax.containers:
    ax.bar_label(container)


plt.show()
```

```python
plt.figure(figsize=(8,5))

ax = sns.countplot(
    data=df,
    x="Age_Group",
    hue="pcos_yn"
)

plt.title("PCOS Distribution Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3)

plt.show()
```
