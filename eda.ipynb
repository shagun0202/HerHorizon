## Feature Engineering

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df['marraige_status_yrs'] = df['marraige_status_yrs'].fillna(df['marraige_status_yrs'].median())
```

```python
fast_food_mode = df['fast_food_yn'].mode()[0]
df['fast_food_yn'] = df['fast_food_yn'].fillna(fast_food_mode)
```

```python
print(df.columns.tolist())
```

```python
df.head()
```

```python
df = df.drop(columns=["sl_no", "patient_file_no"])
```

```python
df.head()
```

```python
df[["weight_kg", "heightcm", "bmi"]].head()
```

```python
df["Age_Group"] = pd.cut(
    df["age_yrs"],
    bins=[0, 18, 25, 35, 100],
    labels=["<18", "18-25", "26-35", "35+"]
)
```

```python
df["Age_Group"].value_counts()
```

```python
df["Long_Cycle"] = (
    df["cycle_lengthdays"] > 35
).astype(int)
```

```python
df.columns.tolist()
```

```python
symptom_cols = [
    "hair_growthyn",
    "pimplesyn",
    "hair_lossyn",
    "weight_gainyn",
    "skin_darkening_yn"
]

df["Symptom_Score"] = df[symptom_cols].sum(axis=1)
```

```python
lifestyle_cols = [
    "fast_food_yn",
    "regexerciseyn"
]
```

```python
for col in lifestyle_cols:
    print(col)
    print(df[col].value_counts())
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="cycleri")

plt.title("Distribution of Menstrual Cycle")
plt.xlabel("Menstrual Cycle")
plt.ylabel("Number of Patients")

plt.show()
```

```python
sns.countplot(
    data=df,
    x='cycleri',
    hue='pcos_yn'
)

plt.title('Menstrual Cycle Regularity vs PCOS')
plt.xlabel('Menstrual Cycle')
plt.ylabel('Number of Patients')
plt.show()
```
