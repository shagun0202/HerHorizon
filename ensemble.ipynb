```python
target = "pcos_yn"
```

```python
X = df.drop(columns=[target])
y = df[target]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from imblearn.pipeline import Pipeline as ImbPipeline
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler ,FunctionTransformer
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

```python
numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    
    ("to_string", FunctionTransformer(
        lambda x: x.astype(str),
        feature_names_out="one-to-one"
    )),
    
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)
```

```python
print("Processed X_train:", X_train_processed.shape)
print("Processed X_test :", X_test_processed.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

print("Number of processed features:", len(feature_names))
```

```python
print(feature_names)
```

```python
X_train_processed = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

X_test_processed = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)
```

## SMOTE

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(
    random_state=42
)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train_processed,
    y_train
)
```

```python
print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(pd.Series(y_train_smote).value_counts())
```

## Random Forest Hyperparameter Tuning

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.ensemble import VotingClassifier

from xgboost import XGBClassifier

# Hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV
```

```python
# Evaluation
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

# SHAP
import shap
```

```python
rf_params = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [None, 5, 10, 15, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", "log2"]
}
```

```python
rf_search = RandomizedSearchCV(
    estimator=RandomForestClassifier(
        random_state=42,
        n_jobs=-1
    ),
    
    param_distributions=rf_params,
    n_iter=20,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_search.fit(
    X_train_smote,
    y_train_smote
)
```

```python
print("Best RF parameters:")
print(rf_search.best_params_)

best_rf = rf_search.best_estimator_
```

## XGBoost Hyperparameter Tuning

```python
xgb_params = {
    "n_estimators": [100, 200, 300, 500],
    "max_depth": [3, 4, 5, 6],
    "learning_rate": [0.01, 0.03, 0.05, 0.1],
    "subsample": [0.7, 0.8, 0.9, 1.0],
    "colsample_bytree": [0.7, 0.8, 0.9, 1.0],
    "min_child_weight": [1, 3, 5]
}
```

```python
xgb_search = RandomizedSearchCV(
    estimator=XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    ),
    
    param_distributions=xgb_params,
    n_iter=25,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

xgb_search.fit(
    X_train_smote,
    y_train_smote
)
```

```python
print("Best XGBoost parameters:")
print(xgb_search.best_params_)

best_xgb = xgb_search.best_estimator_
```

## AdaBoost Hyperparameter Tuning

```python
ada_params = {
    "n_estimators": [50, 100, 200, 300, 500],
    "learning_rate": [0.01, 0.05, 0.1, 0.5, 1.0]
}
```

```python
ada_search = RandomizedSearchCV(
    estimator=AdaBoostClassifier(
        random_state=42
    ),
    
    param_distributions=ada_params,
    n_iter=15,
    scoring="f1",
    cv=5,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

ada_search.fit(
    X_train_smote,
    y_train_smote
)
```

```python
print("Best AdaBoost parameters:")
print(ada_search.best_params_)

best_ada = ada_search.best_estimator_
```

## FINAL ENSEMBLE

```python
final_ensemble = VotingClassifier(
    estimators=[
        ("random_forest", best_rf),
        ("xgboost", best_xgb),
        ("adaboost", best_ada)
    ],
    voting="soft"
)
```

```python
final_ensemble.fit(
    X_train_smote,
    y_train_smote
)

print("Final Ensemble trained successfully!")
```

```python
y_pred = final_ensemble.predict(
    X_test_processed
)

y_probability = final_ensemble.predict_proba(
    X_test_processed
)[:, 1]
```

## EVALUATION

```python
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_probability)

print("===== FINAL ENSEMBLE RESULTS =====")

print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))
print("ROC-AUC  :", round(roc_auc, 4))
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

## CONFUSION MATRIX

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred
)

plt.title("PCOS - Ensemble Confusion Matrix")
plt.show()
```

## Comparison between Individual model vs Ensemble

```python
models = {
    "Random Forest": best_rf,
    "XGBoost": best_xgb,
    "AdaBoost": best_ada,
    "Ensemble": final_ensemble
}

results = []

for name, model in models.items():

    pred = model.predict(X_test_processed)

    if hasattr(model, "predict_proba"):
        prob = model.predict_proba(X_test_processed)[:, 1]
    else:
        prob = None

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred),
        "Recall": recall_score(y_test, pred),
        "F1": f1_score(y_test, pred),
        "ROC-AUC": roc_auc_score(y_test, prob)
    })

results_df = pd.DataFrame(results)

display(results_df)
```

## SHAP 

```python
import shap
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
feature_names = X_train_smote.columns.tolist()
```

```python
print(feature_names)
```

```python
def ensemble_predict(X):

    # Convert NumPy array back to DataFrame
    X = pd.DataFrame(
        X,
        columns=feature_names
    )

    # Probability of class 1 = PCOS
    return final_ensemble.predict_proba(X)[:, 1]
```

```python
background = shap.sample(
    X_train_smote,
    30,
    random_state=42
)

print("Background shape:", background.shape)
```

```python
background = shap.sample(
    X_train_smote,
    30,
    random_state=42
)

print("Background shape:", background.shape)
```

```python
explainer_ensemble = shap.KernelExplainer(
    ensemble_predict,
    background.values
)

print("Ensemble SHAP explainer created successfully!")
```

```python
X_test_sample = X_test_processed.iloc[:10].copy()

print("Test samples:", X_test_sample.shape)
```

```python
shap_values_ensemble = explainer_ensemble.shap_values(
    X_test_sample.values,
    nsamples=100
)

print(
    "SHAP values shape:",
    np.array(shap_values_ensemble).shape
)
```

```python
shap.summary_plot(
    shap_values_ensemble,
    X_test_sample,
    feature_names=feature_names
)
```

```python
patient_index = 0

shap.force_plot(
    explainer_ensemble.expected_value,
    shap_values_ensemble[patient_index],
    X_test_sample.iloc[patient_index],
    feature_names=feature_names,
    matplotlib=True
)

plt.show()
```

```python
shap_explanation = shap.Explanation(
    values=shap_values_ensemble,
    base_values=np.repeat(
        explainer_ensemble.expected_value,
        len(X_test_sample)
    ),
    data=X_test_sample.values,
    feature_names=feature_names
)

shap.plots.beeswarm(
    shap_explanation,
    max_display=20
)
```
